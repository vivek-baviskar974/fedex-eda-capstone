# **Project Summary -**


This dataset is a FedEx logistics dataset that tracks pharmaceutical shipments across multiple countries and regions. It contains 33 columns capturing key information such as shipment mode, purchase orders, vendor agreements, delivery schedules, and product-specific details including medicine type, dosage, and freight cost. The dataset serves as the foundation for analyzing delivery performance and identifying inefficiencies in the global health supply chain.

These are pharmaceutical sector logistics shipments that are distributed across multiple countries, many of which are developing nations. In these shipments, time is a very crucial factor for the patient, because the need for medicine supplies is critical. If shipments do not arrive on schedule, it may cause severe consequences or even death for the patient. These shipments contain ARV (antiretroviral) drugs and medical supplies that help suppress the HIV virus and keep patients alive.

Despite the critical nature of these shipments, delays are a recurring problem across multiple countries and transport modes. The wrong choice of shipment mode, country-specific logistical challenges, and poor planning between purchase orders and scheduled delivery dates are key factors contributing to these delays. These delays are not just operational failures — they directly impact patients whose lives depend on receiving medication on time. Proper identification of the root causes of these delays will help minimize freight costs and improve overall shipment timing.

This analysis aims to identify the root causes of shipment delays by examining multiple contributing factors across the dataset. Key areas of focus include the gap between the Purchase Order date and the Scheduled Delivery Date, the impact of shipment mode on delivery performance, country-wise delay patterns, managing team efficiency, vendor INCO terms, and the relationship between shipment weight and insurance costs. By examining these factors, this analysis will determine what is causing delays, where costs can be reduced, and how the overall logistics system can be improved.

This analysis is conducted using Python as the primary programming language, executed on Google Colab Notebook. Libraries such as Pandas and NumPy are used to load, clean, and explore the dataset, while Matplotlib, Seaborn, and Plotly are used to create visualizations that uncover patterns and trends. The analytical approach follows a structured method — starting with Univariate Analysis (single variable), moving to Bivariate Analysis (two variables), and then Multivariate Analysis (multiple variables together) — to progressively build a deeper understanding of the data.

The findings of this analysis will enable stakeholders to make informed, data-driven decisions across multiple areas of the logistics operation. They will be able to identify which transport modes and vendor INCO terms lead to the best delivery performance, and which managing teams are consistently meeting scheduled delivery dates. Country-wise delay patterns will help in planning smarter shipment routes, while the relationship between shipment weight and freight cost will support better cost management. Ultimately, improving the efficiency of this supply chain means ensuring that life-saving ARV medicines and medical supplies reach patients on time — reducing the risk of severe health consequences and saving lives.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Find all values that cannot be converted to a number
mask = pd.to_numeric(df['Weight (Kilograms)'],errors='coerce').isna()
non_numeric = df['Weight (Kilograms)'][mask].unique()
print (non_numeric)

print (df['Weight (Kilograms)'][mask].value_counts())
````

````python
#Fix Weight (Kilograms) column
df ['Weight (Kilograms)'] = pd.to_numeric(
     df ['Weight (Kilograms)']
     .astype(str)            # changing column datatypes to string
     .str.replace('Kg','',regex=False)  # remove 'Kg' from values
     .str.strip (),           #remove trailing spaces from all rows
     errors='coerce'         # convert remaining non-numeric to NaN
)

#verify
print (df['Weight (Kilograms)'].dtype)
print (df['Weight (Kilograms)'].head (10))
````

````python
# Find all values that cannot be converted to a number
mask = pd.to_numeric(df['Freight Cost (USD)'],errors='coerce').isna()
non_numeric = df['Freight Cost (USD)'][mask].unique()
print (non_numeric)

print (df['Freight Cost (USD)'][mask].value_counts())
````

````python
# Fix Freight Cost (USD) column
df['Freight Cost (USD)'] = pd.to_numeric (
    df['Freight Cost (USD)'],
    errors= 'coerce'
)
#verify
print (df['Freight Cost (USD)'].dtype)
print (df['Freight Cost (USD)'].head (10))
````
