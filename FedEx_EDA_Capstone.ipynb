## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Find all values that cannot be converted to a number
mask = pd.to_numeric(df['Weight (Kilograms)'],errors='coerce').isna()
non_numeric = df['Weight (Kilograms)'][mask].unique()
print (non_numeric)

print (df['Weight (Kilograms)'][mask].value_counts())
````

````python
# Fix Weight (Kilograms) column
df ['Weight (Kilograms)'] = pd.to_numeric(
     df ['Weight (Kilograms)']
     .astype(str)            # changing column datatypes to string
     .str.replace('Kg','',regex=False)  # remove 'Kg' from values
     .str.strip (),           #remove trailing spaces from all rows
     errors='coerce'         # convert remaining non-numeric to NaN
)

# verify
print (df['Weight (Kilograms)'].dtype)
print (df['Weight (Kilograms)'].head (10))
````

````python
# Find all values that cannot be converted to a number
mask = pd.to_numeric(df['Freight Cost (USD)'],errors='coerce').isna()
non_numeric = df['Freight Cost (USD)'][mask].unique()
print (non_numeric)

print (df['Freight Cost (USD)'][mask].value_counts())
````

````python
# Fix Freight Cost (USD) column
df['Freight Cost (USD)'] = pd.to_numeric (
    df['Freight Cost (USD)'],
    errors= 'coerce'
)
# verify
print (df['Freight Cost (USD)'].dtype)
print (df['Freight Cost (USD)'].head (10))
````

````python
#unique values in Dosage column
print (df['Dosage'].unique())
````

````python
# Explore rows where Dosage is missing

print("=== Rows where Dosage is missing ===")
print(df[df['Dosage'].isna()][['Dosage', 'Product Group', 'Dosage Form', 'Shipment Mode']].value_counts())
````

````python
# unique values in Shipment Mode column
print (df['Shipment Mode'].unique())


# Check missing values in Shipment Mode
print ("Missing Shipment Mode count:",df['Shipment Mode'].isnull().sum())
````

````python
# Calculating count and percentage of missing values from Shipment Mode column
print ("Missing Shipment Mode count:",df['Shipment Mode'].isna().sum())
print ("Missing Shipment Mode percentage:",round(360/10324*100,2),'%')
````

````python
# fill missing values from Shipment Mode column using mode
Shipment_Mode_fill = df['Shipment Mode'].mode()[0]
df['Shipment Mode']=df['Shipment Mode'].fillna(Shipment_Mode_fill)

# Verify — Shipment Mode missing count after fill
print ("Shipment Mode missing:",df['Shipment Mode'].isnull().sum())

# Value counts — Shipment Mode before fill
print(df['Shipment Mode'].value_counts())

# most frequent shipment mode (mode value used to fill missing entries)
print("Mode used for Shipment Mode:", Shipment_Mode_fill)
````

````python
# unique values in Line Item Insurance (USD) column
print (df['Line Item Insurance (USD)'].unique())

# Check missing values in Line Item Insurance
print ("Line Item Insurance missing:",df['Line Item Insurance (USD)'].isnull().sum())
````

````python
# replace nan with 0
df['Line Item Insurance (USD)'] = df['Line Item Insurance (USD)'].fillna(0)

# Verify — Line Item Insurance missing count after fill
print ("Line Item Insurance missing:",df['Line Item Insurance (USD)'].isnull().sum())
````

````python
# convert date column to datetime
date_columns = [
    'PQ First Sent to Client Date',
    'PO Sent to Vendor Date',
    'Scheduled Delivery Date',
    'Delivered to Client Date',
    'Delivery Recorded Date'
]

for col in date_columns:
  df[col]= pd.to_datetime(df[col],errors='coerce',format='mixed')


# verify - date column
for col in date_columns:
  print (col,":",df[col].dtype)
````

````python
# Create delay columns
df['Delay_Days'] = (df['Delivered to Client Date'] - df['Scheduled Delivery Date']).dt.days

# Creates binary column (0 or 1)
df['Is_Late'] = (df['Delay_Days'] > 0).astype(int)

# Verifies both new columns exist and look correct
print (df[['Delivered to Client Date','Scheduled Delivery Date','Delay_Days','Is_Late']].head ())
````

````python
# Check for Nan in Delay_Day
print (df['Delay_Days'].isna().sum())

# Check distribution of Delay_Days values
print (df['Delay_Days'].value_counts().head())
````

````python
# lead time = days between PO sent and scheduled delivery
df['lead_time_days'] = (df['Scheduled Delivery Date'] - df['PO Sent to Vendor Date']).dt.days

# verify lead time distribution
df['lead_time_days'].describe()
````

````python
# Group by shipment Mode and convert rows to columns
mode_delivery = df.groupby('Shipment Mode')['Is_Late'].value_counts().unstack(fill_value=0)

#repalce columns name to redable names
mode_delivery.columns =['On-Time','Late']

# Display the  On-Time and Late count for each shipment mode
print(mode_delivery)
````

````python
# Chart 2 — On-Time vs Late Deliveries by Shipment Mode (Grouped Bar Chart)

fig, ax = plt.subplots(figsize=(10, 6))                              # set canvas size

x = np.arange(len(mode_delivery.index))                             # x positions for each mode
width = 0.35                                                         # width of each bar

bars1 = ax.bar(x - width/2, mode_delivery['On-Time'], width,        # on-time bars, shifted left
               label='On-Time', color='steelblue')

bars2 = ax.bar(x + width/2, mode_delivery['Late'], width,           # late bars, shifted right
               label='Late', color='tomato')

ax.set_title('On-Time vs Late Deliveries by Shipment Mode', fontsize=14)  # chart title
ax.set_xlabel('Shipment Mode')                                            # x axis label
ax.set_ylabel('Number of Shipments')                                      # y axis label
ax.set_xticks(x)                                                          # place ticks at mode positions
ax.set_xticklabels(mode_delivery.index)                                   # replace numbers with mode names
ax.legend()                                                               # show color legend

plt.show()                                                                # Display chart
````

#### Chart - 3

````python
# Filter countries with 100+ shipments for reliable analysis
country_counts = df.groupby('Country')['Is_Late'].count()
valid_countries = country_counts[country_counts >= 100].index

# Calculate late rate only for valid countries
country_late = df[df['Country'].isin(valid_countries)].groupby('Country')['Is_Late'].mean() * 100

# Sort from highest to lowest delay rate
country_late = country_late.sort_values(ascending=True)

print(round (country_late,1))
````

````python
# Chart 3 — Late Delivery Rate by Country (100+ Shipments Only)(Horizontal Bar Chart)

fig, ax = plt.subplots(figsize=(13, 6))                              # taller canvas for country names

sns.barplot(x=country_late.values,                                   # x = delay rate %
            y=country_late.index,                                    # y = country names
            hue=country_late.index,                                  # hue mapped to y variable
            palette='Reds',                                          # red color scale
            legend=False,                                            # hide legend
            ax=ax)                                                   # draw on this canvas

ax.set_title('Late Delivery Rate by Country (100+ Shipments Only)', fontsize=14)  # chart title
ax.set_xlabel('Late Delivery Rate (%)')                              # x axis label
ax.set_ylabel('Country')                                             # y axis label

for i, value in enumerate(country_late.values):                      # loop through each bar
    ax.text(value + 0.2, i, f'{value:.1f}%', va='center', fontsize=9)  # add % label at bar end

plt.show()                                                           # Display chart
````

##### 1. Why did you pick the specific chart?

I chose a horizontal bar chart because country names like 'Congo, DRC' and 'Côte d'Ivoire' are long and would overlap each other on a vertical chart, making it unreadable. A single bar per country is sufficient because I am only measuring one metric  the late delivery rate (%)  with no comparison between two groups.

The chart shows that Congo, DRC has the highest late delivery rate at 24.9%, meaning nearly 1 in 4 shipments arrive late. Vietnam has the lowest rate at just 0.9%. A clear pattern emerges  11 out of 16 countries above 10% delay rate are African nations, suggesting that logistics operations in African routes may face structural challenges related to shipment mode selection or route management.

#### Chart - 4

````python
# Chart 4 —  On-Time Delivery Frequency by Shipment Mode (pie chart)

modes = df['Shipment Mode'].dropna().unique()             # unique modes, drop nulls
fig, axes = plt.subplots(1, len(modes), figsize=(16, 6))  # one subplot per mode
colors = ['#2ecc71', '#e74c3c']                           # green=on-time, red=late

for i, mode in enumerate(modes):
    mode_data = df[df['Shipment Mode'] == mode]['Is_Late'] \
                .map({1: 'Late', 0: 'On-Time'}) \
                .value_counts()                           # count on-time vs late per mode

    axes[i].pie(mode_data, labels=mode_data.index,
                autopct='%1.1f%%', colors=colors, startangle=90)  # draw pie
    axes[i].set_title(f'{mode}', fontsize=13, fontweight='bold')   # mode as title

plt.suptitle('Late Delivery Rate (%) Within Each Shipment Mode',
             fontsize=15, fontweight='bold')              # master title
plt.show()                                                # Display Chart
````

To compare all 4 modes simultaneously, multiple pie charts were plotted as subplots one per mode.

#### Chart - 6

````python
# Chart 6 — INCO Term vs Delivery Performance (Stacked Bar Chart)

df['on_time'] = df['Delivered to Client Date'] <= df['Scheduled Delivery Date']  # True = on time

inco_perf = df.groupby(['Vendor INCO Term', 'on_time']).size().reset_index(name='Count')  # count per group

inco_perf['Status'] = inco_perf['on_time'].map({True: 'On Time', False: 'Late'})  # readable labels

# calculate percentage for tooltip
total = inco_perf.groupby('Vendor INCO Term')['Count'].transform('sum')  # total per INCO term
inco_perf['Percentage'] = (inco_perf['Count'] / total * 100).round(2)  # late % per term

fig = px.bar(
    inco_perf,
    x='Vendor INCO Term',              # x axis
    y='Count',                         # y axis = shipment count
    color='Status',                    # color by on time or late
    width=1300,                        # chart width in pixels
    height=600,                        # chart height in pixels
    barmode='stack',                   # stacked bars
    color_discrete_map={'On Time': '#2ecc71', 'Late': '#e74c3c'},  # green and red
    hover_data={'Percentage': True, 'Count': True},  # show % and count on hover
    title='INCO Term vs Delivery Performance'
)

fig.update_layout(
    xaxis_title='Vendor INCO Term',    # x label
    yaxis_title='Number of Shipments', # y label
    legend_title='Delivery Status'     # legend title
)

fig.show()                             # display chart
````

##### 1. Why did you pick the specific chart?

A stacked bar chart was selected because it effectively displays multiple categories  in this case, Vendor INCO Terms  while also showing the proportion of on-time and late shipments within each category. A pie chart was not suitable here as the high number of INCO term categories would make it cluttered and difficult to read for stakeholders. A line chart was also not appropriate because this data does not involve a continuous time period. The stacked bar chart clearly communicates delivery performance across all INCO terms in a single, readable visual.

The chart reveals that N/A From RDC has the highest late shipment proportion at 17.15%, however it does not correspond to any standard international INCO term and likely represents internal distribution center transfers, making it unsuitable for direct comparison. #### Chart - 15 - Pair Plot

````python
#  CHART 15 — Pair Plot

pair_cols = df[[                                         # select columns
    'Freight Cost (USD)',
    'Line Item Quantity',
    'Line Item Value',
    'Line Item Insurance (USD)',
    'Weight (Kilograms)'
]].rename(columns={                                      # short display names
    'Freight Cost (USD)': 'Freight Cost',
    'Line Item Quantity': 'Quantity',
    'Line Item Value': 'Item Value',
    'Line Item Insurance (USD)': 'Insurance',
    'Weight (Kilograms)': 'Weight (Kg)'
})

def smart_fmt(x, pos):                                   # define format function
    if x >= 1_000_000:
        return f'{x/1_000_000:.0f}M'                     # millions → M
    elif x >= 1_000:
        return f'{x/1_000:.0f}K'                         # thousands → K
    else:
        return str(int(x))                               # small numbers as is

g = sns.pairplot(                                        # draw pair plot
    pair_cols,
    diag_kind='kde',                                     # diagonal = density curve
    height=1.8,                                          # subplot height
    aspect=1.3,                                          # wider than tall
    plot_kws={'alpha': 0.5, 'color': 'steelblue'},       # dot transparency
    diag_kws={'color': 'steelblue'}                      # diagonal colour
)

for ax in g.axes.flatten():                                 # loop all subplots
    ax.xaxis.set_major_formatter(FuncFormatter(smart_fmt))  # apply to x
    ax.yaxis.set_major_formatter(FuncFormatter(smart_fmt))  # apply to y

plt.suptitle(                                               # main title
    'Pair Plot — Key Logistics Variables',
    y=1.02, fontsize=14,
)

plt.show()                                                  # display chart
````

##### 1. Why did you pick the specific chart?

A Pair Plot was chosen because it shows the actual scatter of data points between every pair of numeric variables at the same time in one view. Unlike the Correlation Heatmap which gives only a single number per pair, the pair plot shows the shape, direction, and spread of each relationship visually. The diagonal shows the distribution of each individual variable using a KDE curve. This makes it easier to spot patterns, outliers, and whether relationships are linear or not across all variable combinations together.

##### 2. What is/are the insight(s) found from the chart?

As per this visualization, we can clearly see that Item Value and Insurance have a clear upward linear relationship, along with Quantity and Item Value. Weight shows a weak or no clear relationship with other variables  the dots are scattered horizontally with no upward direction. Freight Cost shows a wide scattered pattern against all other variables, meaning it is influenced by many factors and does not follow a single predictable relationship. The Pair Plot visually confirms what the heatmap shows as numbers, but with better visual insights into the shape and spread of each relationship.
